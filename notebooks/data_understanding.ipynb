```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

print("Scraping additional data for Data Enrichment (Scraping and Augmentation Requirement)...")
url = 'https://en.wikipedia.org/wiki/List_of_U.S._states_and_territories_by_population'

headers = {'User-Agent': 'Mozilla/5.0'}
response = requests.get(url, headers=headers)

soup = BeautifulSoup(response.text, 'html.parser')
table = soup.find('table', {'class': 'wikitable'})

data = []
for row in table.find_all('tr'):
    cols = row.find_all(['td', 'th'])
    cols = [ele.text.strip() for ele in cols]
    if cols:
        data.append(cols)

max_len = max(len(r) for r in data)
norm_data = [r + ['']*(max_len - len(r)) for r in data]

scraped_df = pd.DataFrame(norm_data[2:], columns=norm_data[0])

scraped_df = scraped_df.iloc[:, :3]
scraped_df.columns = ['State', 'Population_2024', 'Population_2020']

print("Scraping successful. First 5 rows of scraped dataset:")
display(scraped_df.head())

df['State'] = 'California'
df = pd.merge(df, scraped_df[['State', 'Population_2024']], on='State', how='left')

df['Population_2024'] = df['Population_2024'].str.replace(',', '')
df['Population_2024'] = pd.to_numeric(df['Population_2024'], errors='coerce')
df['Population_2024'] = df['Population_2024'].fillna(df['Population_2024'].median())

print("Data successfully scraped, cleaned, and merged into the main dataframe!")
```

```python
df.to_csv('../data/processed/telecom_enriched.pkl', index=False)
print("Enriched dataset saved successfully!")
```
