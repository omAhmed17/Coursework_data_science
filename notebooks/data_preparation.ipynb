```python
import pandas as pd
import numpy as np
import sys
import os
sys.path.append(os.path.abspath('..'))
```

```python
df = pd.read_pickle(r'../data/processed/telecom_enriched.pkl')
```

```python
df['Offer'] = df['Offer'].fillna(df['Offer'].mode()[0])
df['Internet Type'] = df['Internet Type'].fillna(df['Internet Type'].mode()[0])
df['Customer Satisfaction'] = df['Customer Satisfaction'].fillna(df['Customer Satisfaction'].median())
df['Churn Category'] = df['Churn Category'].fillna('no churn')
df['Churn Reason'] = df['Churn Reason'].fillna('not applicable')
```

```python
cols_to_drop = ['City', 'Zip Code']
df = df.drop(columns=[col for col in cols_to_drop if col in df.columns], errors='ignore')
```

```python
categorical_cols = df.select_dtypes(include=['object']).columns
df_encoded = pd.get_dummies(df, columns=categorical_cols, drop_first=True)
```

```python
df_encoded.head()
```

```python
y = df_encoded['Churn Value'] 

columns_to_drop = [col for col in df_encoded.columns if 'Churn' in col]
print("ALL variables before dropping target variable:", df_encoded.columns.shape[0])
x = df_encoded.drop(columns=columns_to_drop, axis=1)

print("Target variable (y):", y.shape)
print("Feature matrix (X):", x.shape)
```

```python
print(x.select_dtypes(exclude=[np.number]).columns)
```

```python
x.to_pickle('../data/processed/features.pkl')
y.to_pickle('../data/processed/target.pkl')
df_encoded.to_pickle('../data/processed/cleaned_data.pkl')

print("Cleaned data saved successfully!")
```
